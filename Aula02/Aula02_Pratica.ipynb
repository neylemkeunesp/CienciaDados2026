## 1. Documentação Técnica com Markdown e LaTeX

No Jupyter, podemos misturar texto formatado, listas e equações matemáticas com notação LaTeX:

Equação em linha: $f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2}$

Equação em bloco:
$$\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i$$

Nas subseções a seguir, cada elemento da linguagem aparece em dois momentos: a **sintaxe** (o texto que você digita, dentro de um bloco de código) e o **resultado** (como a célula é renderizada). Para ver o texto-fonte de qualquer célula desta aula, dê um **duplo clique** nela; para renderizar de novo, use <kbd>Shift</kbd>+<kbd>Enter</kbd>.

### 1.1 O que é Markdown e como usá-lo no Jupyter

**Markdown** é uma linguagem de marcação leve, criada em 2004 por John Gruber, em que o texto-fonte já é legível mesmo sem renderização. No Jupyter, ele é interpretado quando a célula é do tipo *Markdown* (não *Code*).

| Ação | Atalho (modo de comando, tecla <kbd>Esc</kbd>) |
| :--- | :--- |
| Converter a célula em Markdown | <kbd>M</kbd> |
| Converter a célula em Código | <kbd>Y</kbd> |
| Renderizar a célula | <kbd>Shift</kbd>+<kbd>Enter</kbd> ou <kbd>Ctrl</kbd>+<kbd>Enter</kbd> |
| Editar uma célula Markdown já renderizada | <kbd>Enter</kbd> ou duplo clique |
| Criar célula abaixo / acima | <kbd>B</kbd> / <kbd>A</kbd> |

> **Importante:** existem vários "dialetos" de Markdown. O Jupyter usa um dialeto próximo do **CommonMark** com extensões do **GitHub (GFM)**, como tabelas, texto tachado e listas de tarefas. Recursos como notas de rodapé e emojis por código (`:smile:`) **não** funcionam.

### 1.2 Títulos

Use de um a seis símbolos `#` seguidos de espaço. O nível do título define a hierarquia do documento.

**Sintaxe:**

```markdown

## Título nível 2
### Título nível 3
#### Título nível 4
```

**Resultado:**

💡 **Boa prática:** use um único título de nível 1 por notebook e níveis 2 e 3 para seções. Além de organizar a leitura, os títulos alimentam o **Sumário** do JupyterLab (menu lateral) e a navegação por âncoras.

### 1.10 Fórmulas em LaTeX

O Jupyter renderiza LaTeX com **MathJax**. Use `$...$` para fórmulas **em linha** e `$$...$$` para fórmulas **em bloco**. A tabela abaixo reúne os comandos mais frequentes em Ciência de Dados.

| Elemento | LaTeX | Resultado |
| :--- | :--- | :---: |
| Fração | `\frac{a}{b}` | $\frac{a}{b}$ |
| Potência e índice | `x^2`, `x_i`, `x_i^2` | $x^2$, $x_i$, $x_i^2$ |
| Raiz | `\sqrt{x}`, `\sqrt[3]{x}` | $\sqrt{x}$, $\sqrt[3]{x}$ |
| Letras gregas | `\alpha, \beta, \mu, \sigma, \Sigma` | $\alpha, \beta, \mu, \sigma, \Sigma$ |
| Somatório | `\sum_{i=1}^{n} x_i` | $\sum_{i=1}^{n} x_i$ |
| Produtório | `\prod_{i=1}^{n} x_i` | $\prod_{i=1}^{n} x_i$ |
| Integral | `\int_a^b f(x)\,dx` | $\int_a^b f(x)\,dx$ |
| Barra e chapéu | `\bar{x}`, `\hat{y}` | $\bar{x}$, $\hat{y}$ |
| Relações | `\leq, \geq, \neq, \approx` | $\leq, \geq, \neq, \approx$ |
| Probabilidade | `P(A \mid B)` | $P(A \mid B)$ |
| Texto no meio da fórmula | `x \text{ se } x > 0` | $x \text{ se } x > 0$ |

**Fórmulas em bloco**, com exemplos de estatística:

Variância amostral e desvio padrão:

$$s^2 = \frac{1}{n-1}\sum_{i=1}^{n}\left(x_i - \bar{x}\right)^2, \qquad s = \sqrt{s^2}$$

Teorema de Bayes:

$$P(A \mid B) = \frac{P(B \mid A)\,P(A)}{P(B)}$$

Função definida por partes (`cases`):

$$\mathrm{ReLU}(x) = \begin{cases} x, & \text{se } x > 0 \\ 0, & \text{caso contrário} \end{cases}$$

Várias equações alinhadas pelo sinal de `=` (ambiente `aligned`):

$$\begin{aligned}
\hat{y} &= \beta_0 + \beta_1 x \\
\varepsilon_i &= y_i - \hat{y}_i \\
\mathrm{SSE} &= \sum_{i=1}^{n} \varepsilon_i^2
\end{aligned}$$

Matrizes (`pmatrix` usa parênteses; `bmatrix`, colchetes):

$$A = \begin{pmatrix} 1 & 2 \\ 3 & 4 \end{pmatrix}, \qquad
\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}, \qquad
A\mathbf{x} = \mathbf{b}$$

💡 **Dicas:**
- Use `\left(` e `\right)` para que os parênteses acompanhem a altura da fórmula.
- Use `\,` ou `\quad` para espaçamento e `\text{...}` para texto dentro da fórmula.
- Para escrever o cifrão literal fora de fórmulas (por exemplo, R\$ 10,00), use `\$`.

### 1.11 Gerando Markdown a partir do código Python

Relatórios reprodutíveis combinam texto e resultados. Em vez de copiar números à mão para uma célula Markdown, podemos **gerar o Markdown por código** com `IPython.display.Markdown`. Assim, o texto sempre reflete os dados atuais.

`````python
from IPython.display import Markdown, display

# Dados de exemplo (notas de três provas)
notas = {
    "Ana":   [7.5, 8.0, 9.0],
    "Bruno": [5.0, 5.5, 6.0],
    "Carla": [9.5, 9.0, 10.0],
}

# Monta uma tabela Markdown linha a linha
linhas = ["| Aluno | Média | Situação |", "| :--- | ---: | :---: |"]
for nome, provas in notas.items():
    media = sum(provas) / len(provas)
    situacao = "✅ Aprovado" if media >= 6 else "❌ Reprovado"
    linhas.append(f"| {nome} | {media:.2f} | {situacao} |")

tabela = "\n".join(linhas)

# Texto com fórmula LaTeX preenchida com valores calculados
todas = [n for provas in notas.values() for n in provas]
media_geral = sum(todas) / len(todas)
formula = rf"$\bar{{x}} = \frac{{1}}{{{len(todas)}}} \sum_{{i=1}}^{{{len(todas)}}} x_i = {media_geral:.2f}$"

display(Markdown(f"### Resultado da turma\n\n{tabela}\n\n**Média geral:** {formula}"))
`````

Também é possível escrever Markdown direto em uma célula de **código** com a magic de célula `%%markdown`. Ela é útil para testar rapidamente uma sintaxe sem trocar o tipo da célula:

`````python
%%markdown
**Teste rápido:** esta saída foi gerada por uma célula de *código* com a magic `%%markdown`.

- Item 1
- Item 2 com fórmula $E = mc^2$
`````

### 1.12 Boas práticas e armadilhas comuns

**Estrutura recomendada de um notebook de análise:**

1. **Título e autoria** (nível 1), data e objetivo.
2. **Contexto e pergunta** a ser respondida.
3. **Dados:** origem, licença, dicionário de variáveis.
4. **Metodologia** (com fórmulas, se necessário).
5. **Resultados**, cada um acompanhado de um texto interpretando a figura ou tabela.
6. **Conclusões e limitações.**
7. **Referências e ambiente** (versões das bibliotecas).

**Armadilhas frequentes:**

| Problema | Causa | Solução |
| :--- | :--- | :--- |
| Lista ou tabela não renderiza | Falta linha em branco antes | Inserir uma linha em branco |
| Nome de variável aparece em itálico | Sublinhados `_` interpretados | Colocar entre crases |
| Fórmula não renderiza | `$` desbalanceado ou erro de LaTeX | Conferir chaves `{}` e a abertura/fechamento |
| `$` de valores monetários vira fórmula | Dois `$` na mesma célula | Escapar como `\$` |
| HTML/script não aparece | Notebook não confiável | Preferir Markdown puro |

💡 Escreva o texto pensando em quem vai ler **sem ver o código**: um bom notebook se entende lendo apenas as células Markdown.
