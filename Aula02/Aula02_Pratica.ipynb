### Exercício 2 (Benchmarking com %timeit)
Compare o tempo de execução para criar uma lista de números ao quadrado de 0 a 10.000 usando:
a) Loop `for` tradicional com `.append()`
b) List Comprehension `[i**2 for i in range(10_000)]`


```python
# Teste A: loop tradicional
def cria_loop():
    l = []
    for i in range(10_000):
        l.append(i**2)
    return l

%timeit cria_loop()
```

```python
# Teste B: List comprehension
%timeit [i**2 for i in range(10_000)]
```
