## 3. O que caracteriza uma linguagem? — 8 min

Uma linguagem oferece convenções para expressar computações. Em uma introdução, podemos separar:
- **Sintaxe:** como as expressões e instruções são escritas;
- **Semântica:** o que essas expressões significam;
- **Sistema de tipos:** que categorias de valores existem e quais operações são válidas;
- **Implementação e ecossistema:** compiladores/interpretadores, bibliotecas, ferramentas e comunidade.

**Compilada versus interpretada não é uma classificação absoluta da linguagem.** São estratégias de implementação e execução; sistemas reais podem combinar tradução para código intermediário, interpretação e compilação JIT. Não confunda essa dimensão com paradigma.

Não existe uma linguagem universalmente melhor. A escolha depende, entre outros fatores, da tarefa, desempenho necessário, disponibilidade de bibliotecas, manutenção, interoperabilidade e experiência da equipe.

## 4. Paradigmas: diferentes formas de organizar a solução — 12 min

| Abordagem | Ênfase típica | Pergunta-guia | Linguagens associadas (sem exclusividade) |
|---|---|---|---|
| **Imperativa/procedural** | Sequência de instruções, controle de fluxo e estado | Que passos devem ocorrer? | C, Fortran; também Python |
| **Orientada a objetos** | Objetos que agrupam dados e comportamentos | Que entidades e responsabilidades existem? | Java, C++, Python |
| **Funcional** | Composição de funções e transformação de valores | Que resultado produz esta composição? | Haskell, Lisp; também Python, R |
| **Declarativa** | Descrição do resultado ou das relações desejadas | O que quero obter? | SQL, Prolog |

Estas categorias são lentes didáticas, não caixas rígidas. Uma linguagem pode oferecer construções associadas a vários paradigmas; um programa também pode combinar estilos. A mesma linguagem pode ser usada de modos diferentes.

## 5. Uma tarefa, diferentes maneiras de pensar — 12 min

Considere uma tabela de medições com país e quantidade. A tarefa é **selecionar os países com mais de 200 registros e obter a lista de seus nomes**. Os exemplos abaixo são esquemáticos; servem para comparar ideias, não para executar.

**Imperativo — descreve passos e atualização de uma coleção:**
```text
selecionar = lista vazia
para cada registro na tabela:
    se quantidade do registro > 200:
        acrescentar país a selecionar
```

**Funcional — compõe transformações sobre os registros:**
```text
registros → filtrar quantidade > 200 → extrair país
```

**Declarativo — especifica o resultado desejado em SQL:**
```sql
SELECT pais
FROM catalogo
WHERE quantidade > 200;
```

**Orientado a objetos — modela dados e operações em entidades:** uma aplicação poderia representar cada linha como um objeto `Registro`, com atributos `pais` e `quantidade`; a seleção ainda dependeria de uma operação definida sobre uma coleção desses objetos.

**Compare:** qual trecho explicita mais os passos? Qual descreve mais diretamente a condição do resultado? Há algo importante que todos precisam ter em comum para responder corretamente?

## 8. Paradigma de programação não é modelo de desenvolvimento — 5 min

A aula de 2024 também apresentava os modelos **Catedral** e **Bazar**, associados por Eric S. Raymond a diferentes formas de organizar o desenvolvimento de software. Eles tratam de coordenação, abertura e participação no processo de desenvolvimento — **não** são paradigmas de programação.

Um projeto aberto pode conter código imperativo, funcional e orientado a objetos; um projeto centralizado também pode usar esses mesmos estilos. São dimensões diferentes.
