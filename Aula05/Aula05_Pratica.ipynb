## 1. Métodos de Limpeza de Texto

Dados textuais do mundo real frequentemente contêm espaços excedentes, pontuações indesejadas e variações de caixa.


```python
registro = '   Dr. Ney Lemke;  Física Médica ; UNESP   \n'

# Limpando espaços externos e separando por ponto-e-vírgula
campos = [campo.strip() for campo in registro.split(';')]
print(f'Campos limpos: {campos}')
```

```python
# Transformações de caixa e substituição
frase = 'ciência dos dados na física médica'
print(frase.upper())
print(frase.title())
print(frase.replace('física médica', 'saúde'))
```

## 2. F-Strings Modernas e Fatiamento

As f-strings do Python permitem controle direto de alinhamento e precisão numérica.


```python
pi = 3.1415926535
valor = 1250.75

print(f'Pi formatado: {pi:.3f}')
print(f'Moeda alinhada: R$ {valor:>10.2f}')
```

```python
# Fatiamento [início:fim:passo]
texto = 'Python2026'
print(texto[0:6])   # 'Python'
print(texto[6:])    # '2026'
print(texto[::-1])  # Invertido
```

### Exercício 1 (Higienização de CPFs)
Dada a lista de documentos `[' 123.456.789-00 ', '987.654.321-11\n', '111.222.333-44']`, crie uma nova lista contendo apenas os dígitos numéricos limpos.


```python
docs = [' 123.456.789-00 ', '987.654.321-11\n', '111.222.333-44']
docs_limpos = [d.strip().replace('.', '').replace('-', '') for d in docs]
print(f'CPFs normalizados: {docs_limpos}')
```

### Exercício 2 (Extração de Domínio de E-mail)
Escreva uma função que receba um e-mail (ex: `aluno@unesp.br`) e retorne apenas o domínio (ex: `unesp.br`).


```python
def extrai_dominio(email: str) -> str:
    return email.strip().split('@')[-1]

print(extrai_dominio('ney.lemke@unesp.br'))
```
