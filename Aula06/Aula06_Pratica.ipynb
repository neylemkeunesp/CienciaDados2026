## 1. As Quatro Coleções Nativas de Python

- **Lista (`list`)**: ordenada, mutável, permite duplicatas.
- **Tupla (`tuple`)**: ordenada, imutável (ideal para registros fixos).
- **Conjunto (`set`)**: não-ordenado, mutável, elementos únicos (busca rápida $O(1)$).
- **Dicionário (`dict`)**: pares chave-valor, chaves únicas.


```python
idades = [21, 25, 21, 30, 25, 19, 21]

# Deduplicação instantânea com set
idades_unicas = set(idades)
print(f'Original: {idades}')
print(f'Únicos: {idades_unicas}')
```

## 2. Comprehensions (Listas e Dicionários)

Forma idiomática e de alta performance para mapear e filtrar coleções.


```python
# Quadrados dos números pares de 0 a 10
quadrados_pares = [x**2 for x in range(11) if x % 2 == 0]
print(f'Quadrados pares: {quadrados_pares}')

# Dict comprehension: mapeando palavra -> seu tamanho
palavras = ['dados', 'ciência', 'python', 'análise']
tamanhos = {p: len(p) for p in palavras}
print(f'Mapeamento de tamanhos: {tamanhos}')
```

### Exercício 1 (Contagem de Frequências)
Dada uma lista de diagnósticos clínicos, conte quantas vezes cada diagnóstico aparece utilizando um dicionário.


```python
diagnosticos = ['Gripe', 'Covid', 'Dengue', 'Gripe', 'Covid', 'Gripe', 'Resfriado']

contagem = {}
for diag in diagnosticos:
    contagem[diag] = contagem.get(diag, 0) + 1

print(f'Frequências: {contagem}')
```

### Exercício 2 (Operações com Conjuntos)
Dados dois conjuntos de pacientes que realizaram exames A e B, encontre:
a) Pacientes que fizeram ambos os exames.
b) Pacientes que fizeram apenas o exame A.


```python
exame_a = {'P01', 'P02', 'P03', 'P04', 'P05'}
exame_b = {'P03', 'P05', 'P06', 'P07'}

ambos = exame_a.intersection(exame_b)
apenas_a = exame_a.difference(exame_b)

print(f'Fizeram ambos: {ambos}')
print(f'Apenas exame A: {apenas_a}')
```
