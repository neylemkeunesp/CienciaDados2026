## 1. Tratamento de Dados Ausentes (Missing Values)


```python
import pandas as pd
import numpy as np

df_sujo = pd.DataFrame({
    'id': [101, 102, 103, 104, 105],
    'temperatura': [36.5, np.nan, 38.2, 37.0, np.nan],
    'pressao': [120, 130, np.nan, 125, 118],
    'setor': ['A', 'A', 'B', 'B', 'A']
})

print('Valores nulos por coluna:')
print(df_sujo.isna().sum())
```

```python
# Imputação de missing pela mediana da coluna
mediana_temp = df_sujo['temperatura'].median()
df_limpo = df_sujo.copy()
df_limpo['temperatura'] = df_limpo['temperatura'].fillna(mediana_temp)
df_limpo
```

## 2. Agregações Complexas com GroupBy


```python
# Agrupando por setor e calculando múltiplas métricas com .agg()
resumo_setor = df_limpo.groupby('setor').agg(
    temp_media=('temperatura', 'mean'),
    temp_max=('temperatura', 'max'),
    contagem=('id', 'count')
)
resumo_setor
```

## 3. Junção de Tabelas (Merge)


```python
df_meta = pd.DataFrame({
    'setor': ['A', 'B'],
    'responsavel': ['Dra. Julia', 'Dr. Roberto'],
    'andar': [2, 3]
})

# Inner join relacionando pacientes ao responsável pelo setor
df_completo = pd.merge(df_limpo, df_meta, on='setor', how='left')
df_completo
```
