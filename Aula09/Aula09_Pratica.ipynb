## 1. Series e DataFrame


```python
import pandas as pd

# pd.Series: vetor unidimensional com rótulos
idades = pd.Series([25, 42, 60, 31, 55], index=['P01', 'P02', 'P03', 'P04', 'P05'], name='idade')
print(idades)
print('Índice:', list(idades.index))
print('Valores:', list(idades.values))
```

```python
# pd.DataFrame: tabela de Séries (colunas)
dados = {
    'paciente': ['P01', 'P02', 'P03', 'P04', 'P05'],
    'idade': [25, 42, 60, 31, 55],
    'glicose': [92, 115, 140, 88, 128],
    'grupo': ['Controle', 'Tratamento', 'Tratamento', 'Controle', 'Tratamento']
}

df = pd.DataFrame(dados)
df
```

```python
# Inspeção sumária
df.info()
df.describe()
```

## 2. Leitura de Dados Tabulares (pd.read_csv)

- `pd.read_csv('arquivo.csv')` lê um CSV e devolve um `DataFrame`.
- `df.head()`, `df.shape` e `df.dtypes` resumem a estrutura do dado.


```python
# Carregando um CSV real da pasta da aula
df_ascii = pd.read_csv('dicchar.csv')
df_ascii.head()
```

```python
df_ascii.info()
print('Dimensões:', df_ascii.shape)
print('Tipos:', dict(df_ascii.dtypes))
```

## 3. Indexação com .loc e .iloc

- `.loc`: seleção baseada em nomes de colunas e rótulos de índice.
- `.iloc`: seleção baseada estritamente em índices inteiros posicionais.


```python
# Selecionando linha 0 a 2 e colunas 'paciente' e 'glicose'
print(df.loc[0:2, ['paciente', 'glicose']])

# Selecionando com iloc (posições 0 e 1, colunas 0 e 1)
print(df.iloc[0:2, 0:2])
```

## 4. Filtros Booleanos Combinados (&, |, ~)

- `&` (E) entre duas condições.
- `|` (OU) entre duas condições.
- `~` (NÃO) inverte uma condição.


```python
# & (E): Tratamento E glicose >= 120
mascara_e = (df['grupo'] == 'Tratamento') & (df['glicose'] >= 120)
df[mascara_e]
```

```python
# | (OU): Controle OU glicose >= 120
mascara_ou = (df['grupo'] == 'Controle') | (df['glicose'] >= 120)
df[mascara_ou]
```

```python
# ~ (NÃO): todos exceto Tratamento
mascara_nao = ~(df['grupo'] == 'Tratamento')
df[mascara_nao]

# Combinado com dado real: dígitos ASCII (num de 48 a 57)
df_ascii[(df_ascii['num'] >= 48) & (df_ascii['num'] <= 57)]
```

### Exercício 1 (Cálculo de Novas Colunas)

### Exercício 2 (Filtro Combinado com &)
No DataFrame de pacientes, selecione os que têm idade entre 30 e 55 anos **e** glicose abaixo de 130.


```python
df[(df['idade'] >= 30) & (df['idade'] <= 55) & (df['glicose'] < 130)]
```

### Exercício 3 (Agrupamento Descritivo Simples)

```python
df.groupby('grupo')[['idade', 'glicose']].mean()
```
