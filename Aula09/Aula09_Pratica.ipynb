## 1. Criação e Inspeção de DataFrames


```python
import pandas as pd

# Criando um DataFrame a partir de um dicionário
dados = {
    'paciente': ['P01', 'P02', 'P03', 'P04', 'P05'],
    'idade': [25, 42, 60, 31, 55],
    'glicose': [92, 115, 140, 88, 128],
    'grupo': ['Controle', 'Tratamento', 'Tratamento', 'Controle', 'Tratamento']
}

df = pd.DataFrame(dados)
df
```

```python
# Inspeção sumária
print(df.info())
print('\nEstatísticas Numéricas:')
print(df.describe())
```

## 2. Indexação e Filtragem Booleana

- `.loc`: seleção baseada em nomes de colunas e rótulos de índice.
- `.iloc`: seleção baseada estritamente em índices inteiros posicionais.


```python
# Selecionando linha 0 a 2 e colunas 'paciente' e 'glicose'
print(df.loc[0:2, ['paciente', 'glicose']])

# Selecionando com iloc (posições 0 e 1, colunas 0 e 1)
print(df.iloc[0:2, 0:2])
```

```python
# Filtro booleano: pacientes do grupo 'Tratamento' com glicose >= 120
mascara = (df['grupo'] == 'Tratamento') & (df['glicose'] >= 120)
df[mascara]
```
