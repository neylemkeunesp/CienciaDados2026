## 1. Visualização de Distribuições e Outliers


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

sns.set_theme(style='whitegrid', palette='deep')

df_carros = pd.read_csv('Automobile price data.csv')
df_carros['price'] = pd.to_numeric(df_carros['price'], errors='coerce')
df_carros['horsepower'] = pd.to_numeric(df_carros['horsepower'], errors='coerce')
df_carros_clean = df_carros.dropna(subset=['price', 'horsepower', 'fuel-type', 'body-style'])

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Histograma com curva KDE
sns.histplot(df_carros_clean['price'], kde=True, ax=axes[0], color='#1B4973')
axes[0].set_title('Distribuição de Preços (Assimetria Positiva)')

# Boxplot comparativo por carroceria
sns.boxplot(data=df_carros_clean, x='body-style', y='price', ax=axes[1])
axes[1].set_title('Preço por Tipo de Carroceria')
axes[1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

## 2. Matriz de Correlação e Heatmap


```python
colunas_numericas = ['price', 'horsepower', 'city-mpg', 'highway-mpg', 'curb-weight', 'engine-size']
for col in colunas_numericas:
    df_carros_clean[col] = pd.to_numeric(df_carros_clean[col], errors='coerce')

corr_matrix = df_carros_clean[colunas_numericas].dropna().corr()

plt.figure(figsize=(7, 5))
sns.heatmap(corr_matrix, annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)
plt.title('Matriz de Correlação de Pearson')
plt.tight_layout()
plt.show()
```

### Exercício 1 (Scatterplot Categórico)
Utilize `sns.scatterplot()` para cruzar `engine-size` (eixo X) com `price` (eixo Y), diferenciando as cores (`hue`) pelo tipo de combustível (`fuel-type`).


```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df_carros_clean,
    x='engine-size',
    y='price',
    hue='fuel-type',
    alpha=0.8,
    s=70
)
plt.title('Preço vs Tamanho do Motor por Combustível')
plt.tight_layout()
plt.show()
```
