## 1. A Interface Orientada a Objetos (`plt.subplots`)


```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Dados sintéticos de calibração instrumental
tempo = np.linspace(0, 10, 50)
sinal = 2.5 * np.sin(tempo) + np.random.normal(0, 0.2, size=50)

fig, ax = plt.subplots(figsize=(8, 4), dpi=100)
ax.plot(tempo, sinal, color='#0F2B45', linewidth=2, label='Sinal Medido')
ax.set_title('Resposta Temporal do Sensor', fontsize=14, fontweight='bold', pad=12)
ax.set_xlabel('Tempo (s)', fontsize=12)
ax.set_ylabel('Tensão (mV)', fontsize=12)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(frameon=True)
plt.tight_layout()
plt.show()
```

## 2. Gráficos de Dispersão com Dataset Real


```python
# Carregando o conjunto de dados automotivo
df_carros = pd.read_csv('Automobile price data.csv')

# Limpeza simples de valores não numéricos em 'price' e 'horsepower'
df_carros['price'] = pd.to_numeric(df_carros['price'], errors='coerce')
df_carros['horsepower'] = pd.to_numeric(df_carros['horsepower'], errors='coerce')
df_carros_clean = df_carros.dropna(subset=['price', 'horsepower'])

fig, ax = plt.subplots(figsize=(8, 5))
scatter = ax.scatter(
    df_carros_clean['horsepower'],
    df_carros_clean['price'],
    c=df_carros_clean['city-mpg'],
    cmap='viridis',
    alpha=0.7,
    edgecolors='none'
)
cbar = fig.colorbar(scatter, ax=ax)
cbar.set_label('Consumo Urbano (City MPG)')

ax.set_title('Relação entre Potência (HP) e Preço de Automóveis', fontsize=13)
ax.set_xlabel('Potência (Horsepower)')
ax.set_ylabel('Preço (USD)')
ax.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

### Exercício 1 (Gráfico de Barras Médio)
Crie um gráfico de barras exibindo o preço médio dos automóveis por tipo de combustível (`fuel-type`).


```python
preco_combustivel = df_carros_clean.groupby('fuel-type')['price'].mean()

fig, ax = plt.subplots(figsize=(6, 4))
ax.bar(preco_combustivel.index, preco_combustivel.values, color=['#1B4973', '#0E7C86'], width=0.5)
ax.set_title('Preço Médio por Tipo de Combustível')
ax.set_ylabel('Preço Médio (USD)')
ax.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
