## 1. Tipos de Dados Primitivos e Type Casting

Python é uma linguagem dinamicamente tipada, porém fortemente tipada.


```python
a = 10          # int
b = 3.14159     # float
c = 'UNESP'     # str
d = True        # bool

print(type(a), type(b), type(c), type(d))
```

```python
# Conversão explícita de tipos (Casting)
str_num = '123'
num = int(str_num)
float_num = float(num)
print(f'Valor: {float_num}, Tipo: {type(float_num)}')
```

## 2. Paradigma Imperativo vs Funcional

Vamos calcular a soma dos quadrados dos números pares de 1 a 10 de duas formas.


```python
# Abordagem 1: Imperativa (instruindo o passo a passo com mutação de estado)
soma_imperativa = 0
for x in range(1, 11):
    if x % 2 == 0:
        soma_imperativa += x ** 2

print(f'Soma imperativa: {soma_imperativa}')
```

```python
# Abordagem 2: Funcional (composição de funções puras e imutabilidade)
pares = filter(lambda x: x % 2 == 0, range(1, 11))
quadrados = map(lambda x: x ** 2, pares)
soma_funcional = sum(quadrados)

print(f'Soma funcional: {soma_funcional}')
```

### Exercício 1 (Operadores e Tipagem)
Escreva uma função com type hints que receba um peso em kg (float) e uma altura em metros (float), calcule o IMC e retorne o valor com duas casas decimais.


```python
def calcula_imc(peso: float, altura: float) -> float:
    return round(peso / (altura ** 2), 2)

print(f'IMC: {calcula_imc(70.0, 1.75)}')
```

### Exercício 2 (Estilo Funcional)
Dada a lista de temperaturas em Celsius `[0, 15, 20, 25, 30]`, utilize `map` e uma função lambda para convertê-las para Fahrenheit: $F = C \times 1.8 + 32$.


```python
celsius = [0, 15, 20, 25, 30]
fahrenheit = list(map(lambda c: c * 1.8 + 32, celsius))
print(f'Fahrenheit: {fahrenheit}')
```
