## 1. Aprendizado Supervisionado: Regressão Linear

Vamos prever o preço de um automóvel com base em sua potência (`horsepower`).


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Carregando dados
df = pd.read_csv('Automobile price data.csv')
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['horsepower'] = pd.to_numeric(df['horsepower'], errors='coerce')
df_reg = df.dropna(subset=['price', 'horsepower'])

# Matriz de features X (2D) e vetor target y (1D)
X = df_reg[['horsepower']]
y = df_reg['price']

# Instanciando e treinando o modelo
modelo = LinearRegression()
modelo.fit(X, y)

y_pred = modelo.predict(X)

print(f'Coeficiente Angular (Slope): {modelo.coef_[0]:.2f}')
print(f'Intercepto: {modelo.intercept_:.2f}')
print(f'R² Score: {r2_score(y, y_pred):.3f}')
print(f'RMSE: {np.sqrt(mean_squared_error(y, y_pred)):.2f}')
```

```python
# Visualizando o ajuste
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='#1B4973', alpha=0.6, label='Dados Reais')
plt.plot(X, y_pred, color='#D97706', linewidth=2.5, label='Reta de Regressão')
plt.title('Regressão Linear: Preço vs Potência')
plt.xlabel('Potência (Horsepower)')
plt.ylabel('Preço (USD)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 2. Aprendizado Não-Supervisionado: Clusterização K-Means

Identificando grupos naturais de veículos combinando potência e consumo de combustível.


```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

df_reg['city-mpg'] = pd.to_numeric(df_reg['city-mpg'], errors='coerce')
X_cluster = df_reg[['horsepower', 'city-mpg']].dropna()

# Padronização de escala (essencial para métodos baseados em distância)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)

# Ajuste de K-Means com k=3 clusters
kmeans = KMeans(n_clusters=3, random_state=42, n_init='auto')
X_cluster['cluster'] = kmeans.fit_predict(X_scaled)

plt.figure(figsize=(8, 5))
sns_plot = plt.scatter(X_cluster['horsepower'], X_cluster['city-mpg'], c=X_cluster['cluster'], cmap='tab10', alpha=0.7)
plt.title('Agrupamento K-Means (k=3)')
plt.xlabel('Potência (HP)')
plt.ylabel('Consumo (City MPG)')
plt.grid(True, linestyle=':', alpha=0.6)
plt.colorbar(sns_plot, label='Cluster')
plt.tight_layout()
plt.show()
```
