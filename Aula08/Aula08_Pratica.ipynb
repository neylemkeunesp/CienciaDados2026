## 1. Leitura de Arquivo Texto com Gerenciador de Contexto (`with`)


```python
with open('texto.txt', 'r', encoding='utf-8') as f:
    linhas = [linha.strip() for linha in f.readlines() if linha.strip()]

print(f'Total de linhas lidas: {len(linhas)}')
print(f'Primeira linha: {linhas[0]}')
```

## 2. Leitura Direta de Arquivo Compactado (`.gz`)


```python
import gzip

# Lendo arquivo comprimido da estação de outono
with gzip.open('arq_abril_outono.txt.gz', 'rt', encoding='utf-8') as f:
    primeiras_linhas = [f.readline().strip() for _ in range(5)]

for l in primeiras_linhas:
    print(l)
```

## ✍️ Exercícios Práticos da Aula 08

### Exercício 1 (Parsing de CSV Manual)
Leia o arquivo `matriznum.csv` (delimitado por espaços) usando o módulo `csv` e calcule a soma dos valores da primeira coluna numérica.


```python
import csv

valores_col1 = []
with open('matriznum.csv', 'r', encoding='utf-8') as f:
    leitor = csv.reader(f, delimiter=' ')
    for linha in leitor:
        linha_limpa = [x for x in linha if x]
        if linha_limpa:
            try:
                valores_col1.append(float(linha_limpa[0]))
            except ValueError:
                continue

print(f'Valores coletados: {valores_col1}')
print(f'Soma da primeira coluna: {sum(valores_col1):.2f}')
```

### Exercício 2 (Escrita de Arquivo de Relatório)
Gere um arquivo `relatorio_aula08.txt` salvando a contagem de elementos e a média calculada.


```python
media_col1 = sum(valores_col1) / len(valores_col1) if valores_col1 else 0.0
with open('relatorio_aula08.txt', 'w', encoding='utf-8') as f:
    f.write('Relatório de Processamento - Aula 08\n')
    f.write(f'Quantidade de elementos: {len(valores_col1)}\n')
    f.write(f'Média: {media_col1:.2f}\n')

print('Relatório gerado com sucesso.')
```
