## 1. Iteração Idiomática com `enumerate` e `zip`


```python
pacientes = ['Ana', 'Bruno', 'Carla', 'Daniel']
idades = [24, 31, 28, 45]

# zip para iterar em paralelo
for nome, idade in zip(pacientes, idades):
    print(f'{nome} tem {idade} anos.')
```

```python
# enumerate para obter índice e elemento simultaneamente
for idx, nome in enumerate(pacientes, start=1):
    print(f'Paciente #{idx}: {nome}')
```

## 2. Funções Modulares e Robustas


```python
def estatisticas_basicas(valores: list[float]) -> dict[str, float]:
    """
    Calcula média, mínimo e máximo de uma lista numérica.
    """
    if not valores:
        raise ValueError('A lista não pode estar vazia.')
    
    return {
        'media': sum(valores) / len(valores),
        'minimo': min(valores),
        'maximo': max(valores)
    }

print(estatisticas_basicas([10.5, 20.0, 15.2, 33.1]))
```

### Exercício 1 (Função com Validação e Tratamento de Erros)
Escreva uma função `converte_para_float(lista_strings)` que converta cada elemento para float, ignorando valores que causem `ValueError` (ex: `'N/A'`, `'ausente'`).


```python
def converte_para_float(lista: list[str]) -> list[float]:
    validos = []
    for item in lista:
        try:
            validos.append(float(item))
        except ValueError:
            continue
    return validos

dados_brutos = ['12.5', '30.1', 'N/A', '45.0', 'invalido', '2.8']
print(f'Valores numéricos convertidos: {converte_para_float(dados_brutos)}')
```

### Exercício 2 (Classificador Simples com if/elif/else)
Crie uma função que classifique o nível de glicose em jejum: `< 100` (Normal), `100 a 125` (Pré-diabetes), `>= 126` (Diabetes).


```python
def classifica_glicose(valor: float) -> str:
    if valor < 100:
        return 'Normal'
    elif valor <= 125:
        return 'Pré-diabetes'
    else:
        return 'Diabetes'

print(classifica_glicose(95))
print(classifica_glicose(110))
print(classifica_glicose(140))
```
