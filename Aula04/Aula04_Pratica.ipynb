## 1. Navegação e Inspeção de Arquivos

Antes de carregar conjuntos de dados de dezenas de gigabytes no Python, cientistas de dados inspecionam sua estrutura com ferramentas Unix.


```python
# Verificando os arquivos da pasta da aula de manipulação (Aula08)
!ls -lh ../Aula08/
```

```python
# Inspecionando as primeiras 5 linhas de um arquivo texto
!head -n 5 ../Aula08/texto.txt
```

```python
# Contando linhas, palavras e caracteres (wc)
!wc -l ../Aula08/texto.txt
```

## 2. Filtragem e Pipelines Unix

Podemos combinar comandos com pipes (`|`) para filtrar e contar ocorrências.


```python
# Filtrando linhas que contêm a palavra 'dados' (case insensitive)
!grep -i 'dados' ../Aula08/texto.txt | wc -l
```
