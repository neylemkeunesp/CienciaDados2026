## 2. Modelando Dados: O Catálogo de Leporello (Don Giovanni)

Vamos modelar o catálogo de conquistas de Don Giovanni descrito por Leporello como uma lista de registros (dicionários).


```python
# Dados de registro por região/país
catalogo = [
    {'pais': 'Itália', 'quantidade': 640},
    {'pais': 'Alemanha', 'quantidade': 231},
    {'pais': 'França', 'quantidade': 100},
    {'pais': 'Turquia', 'quantidade': 91},
    {'pais': 'Espanha', 'quantidade': 1003}
]

print('Catálogo de Conquistas:')
for reg in catalogo:
    print(f"{reg['pais']:<10}: {reg['quantidade']:>5}")
```

## 3. Agregações e Estatísticas Básicas

Calculando o total de registros e encontrando a região com maior contagem.


```python
total = sum(reg['quantidade'] for reg in catalogo)
maior_reg = max(catalogo, key=lambda x: x['quantidade'])

print(f'Total acumulado: {total}')
print(f"País com mais registros: {maior_reg['pais']} com {maior_reg['quantidade']}")
```

## ✍️ Exercícios Práticos da Aula 01

### Exercício 1 (Fixação)
Calcule a média de registros por país no catálogo de Leporello utilizando apenas funções embutidas de Python (`sum`, `len`).


```python
# Seu código aqui
media = total / len(catalogo)
print(f'Média por país: {media:.2f}')
```

### Exercício 2 (Aplicação)
Crie uma lista contendo apenas os países que possuem mais de 200 registros.


```python
# Seu código aqui
paises_mais_200 = [reg['pais'] for reg in catalogo if reg['quantidade'] > 200]
print(f'Países com mais de 200 registros: {paises_mais_200}')
```

### Exercício 3 (Desafio)
Calcule a porcentagem que a Espanha representa sobre o total de registros do catálogo.


```python
# Seu código aqui
espanha_qtd = next(reg['quantidade'] for reg in catalogo if reg['pais'] == 'Espanha')
percentual_espanha = (espanha_qtd / total) * 100
print(f'A Espanha representa {percentual_espanha:.1f}% do catálogo.')
```
