# Ciência dos Dados I — Aula 01
## Roteiro prático: explorar as fontes com NotebookLM

**Tempo estimado:** 60 minutos · **Formato:** individual e em duplas

### Objetivos
- Encontrar informações no material e conferir sua origem.
- Formular e aprimorar prompts para investigar as fontes.
- Organizar exemplos históricos de coleta e análise de dados.
- Interpretar números e reconhecer limites do que se pode concluir.



## Como trabalhar

1. Abra o NotebookLM da aula e confirme que os slides/PDF da Aula 01 estão entre as fontes.
2. Em cada atividade, **escreva seu próprio prompt antes de consultar o NotebookLM**. Não há prompts prontos neste roteiro.
3. Registre o prompt exatamente como foi enviado. Leia a resposta e confira as referências no material.
4. Se a resposta estiver vaga, incompleta ou sem apoio na fonte, reformule o prompt e registre a nova versão.
5. Escreva suas respostas neste notebook, editando esta célula ou acrescentando células de texto/Markdown.

### Verificação do prompt
Antes de enviar, confira se seu prompt deixa claro:
- qual tarefa você quer que o NotebookLM realize;
- quais fontes ou limites deve respeitar;
- que tipo de resposta você espera;
- como você poderá conferir a resposta no material.

> Uma resposta da IA é uma pista para consultar a fonte, não uma fonte por si só.
